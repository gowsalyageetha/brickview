```python
import sqlite3

#connect/create database
conn = sqlite3.connect("brickview_realestate.db")
cursor = conn.cursor()
```

```python
create_listings_table = """
CREATE TABLE IF NOT EXISTS listings (
    listing_id INTEGER PRIMARY KEY,
    address TEXT,
    city TEXT,
    state TEXT,
    zip TEXT,
    property_type TEXT,
    price REAL,
    area_sqft REAL,
    agent_id INTEGER,
    listed_date DATE
    
    );
"""
cursor.execute(create_listings_table)
conn.commit()
```

```python
cursor.execute("PRAGMA table_info(listings);")
columns =cursor.fetchall()

for col in columns:
    print(col)
```

```python
    
create_property_attributes_table = """
CREATE TABLE IF NOT EXISTS property_attributes (
attribute_id INTEGER PRIMARY KEY,
listing_id INTEGER,
bedrooms INTEGER,
bathrooms INTEGER,
floor_number INTEGER,
total_floors INTEGER,
year_built INTEGER,
is_rented BOOLEAN,
tenant_count INTEGER,
furnishing_status TEXT,
metro_distance_km REAL,
parking_available BOOLEAN,
power_backup BOOLEAN,
FOREIGN KEY (listing_id) REFERENCES listings(listing_id)
);
"""
cursor.execute(create_property_attributes_table)
conn.commit()
```

```python
#checking table was created
cursor.execute("PRAGMA table_info(property_attributes);")
columns =cursor.fetchall()

for col in columns:
    print(col)
```

```python
creat_agents_table ="""
CREATE TABLE IF NOT EXISTS agents(
agent_id INTEGER PRIMARY KEY,
name TEXT,
city TEXT,
contact TEXT,
commission_rate REAL,
deals_closed INTEGER,
rating REAL,
experience_years INTEGER,
avg_closing_days INTEGER,
FOREIGN KEY (agent_id) REFERENCES agents(agent_id)
);
"""
cursor.execute(creat_agents_table)
conn.commit()



```

```python
create_sales_table ="""
CREATE TABLE IF NOT EXISTS sales(
sale_id INTEGER PRIMARY KEY,
listing_id INTEGER,
sale_date DATE,
sale_price REAL,
days_on_market INTEGER,
FOREIGN KEY (listing_id) REFERENCES listings(listing_id)
);
"""
cursor.execute(create_sales_table)
conn.commit()
```

```python
create_buyers_table ="""
CREATE TABLE IF NOT EXISTS buyers(
buyer_id INTEGER PRIMARY KEY,
sale_id INTEGER,
buyer_type TEXT,
payment_mode TEXT,
loan_taken BOOLEAN,
loan_provider TEXT,
loan amount REAL,
FOREIGN KEY (sale_id) REFERENCES sales(sale_id)
);
"""
cursor.execute(create_buyers_table)
conn.commit()
```

```python
listings_df.to_sql("listings", conn, if_exists="replace", index=False)
property_attributes_df.to_sql("property_attributes", conn, if_exists="replace", index=False)
agents_df.to_sql("agents", conn, if_exists="replace", index=False)
sales_df.to_sql("sales", conn, if_exists="replace", index=False)
buyers_df.to_sql("buyers", conn, if_exists="replace",index=False)
```

```python
#Q1. What is the average listing price by city?
query = """
SELECT city, ROUND(AVG(price)) AS avg_price
from listings
GROUP BY city
ORDER BY avg_price desc;
"""
df_avg_price_city =pd.read_sql(query, conn)
print(df_avg_price_city)
```

```python
#Q2. Average price per square foot by property type
query = """
SELECT Property_Type,
       ROUND(AVG(Price / Sqft), 2) AS Avg_Price_Per_Sqft
FROM listings
GROUP BY Property_Type
ORDER BY Avg_Price_Per_Sqft DESC;
"""

df_avg_price_listings = pd.read_sql(query, conn)
print(df_avg_price_listings)
```

```python
#Q3. Average price by furnishing status
query="""
SELECT
p.Furnishing_Status,
ROUND(AVG(l.Price),2) AS Avg_Price
FROM listings l
JOIN property_attributes p
ON l.Listing_ID=p.Listing_ID
GROUP BY p.Furnishing_Status;
"""

df_furnishing=pd.read_sql(query,conn)
print(df_furnishing)
```

```python
#Q4.How does furnishing status impact property prices?
query="""
SELECT
p.Furnishing_Status,
ROUND(AVG(l.Price),2) AS Avg_Price
FROM listings l
JOIN property_attributes p
ON l.Listing_ID=p.Listing_ID
GROUP BY p.Furnishing_Status;
"""

df_furnishing=pd.read_sql(query,conn)
print(df_furnishing)
```

```python
#Q5. Do properties closer to metro stations command higher prices?
query="""
SELECT
CASE
WHEN Metro_Distance_Km<1 THEN 'Less than 1 KM'
WHEN Metro_Distance_Km BETWEEN 1 AND 3 THEN '1-3 KM'
ELSE 'More than 3 KM'
END AS Metro_Distance,

ROUND(AVG(l.Price),2) AS Avg_Price

FROM listings l
JOIN property_attributes p
ON l.Listing_ID=p.Listing_ID

GROUP BY Metro_Distance;
"""

df_metro=pd.read_sql(query,conn)
print(df_metro)
```

```python
#Q6.Are rented properties priced differently?
query="""
SELECT
Is_Rented,
ROUND(AVG(l.Price),2) AS Avg_Price

FROM listings l
JOIN property_attributes p
ON l.Listing_ID=p.Listing_ID

GROUP BY Is_Rented;
"""

df_rented=pd.read_sql(query,conn)
print(df_rented)
```

```python
#Q7. How do bedrooms and bathrooms affect pricing?
query="""
SELECT
Bedrooms,
Bathrooms,
ROUND(AVG(l.Price),2) AS Avg_Price

FROM listings l
JOIN property_attributes p
ON l.Listing_ID=p.Listing_ID

GROUP BY Bedrooms,Bathrooms

ORDER BY Bedrooms,Bathrooms;
"""

df_bedroom=pd.read_sql(query,conn)
print(df_bedroom)
```

```python
#Q8.Do properties with parking and power backup have higher prices?
query="""
SELECT
Parking_Available,
Power_Backup,
ROUND(AVG(l.Price),2) AS Avg_Price

FROM listings l
JOIN property_attributes p
ON l.Listing_ID=p.Listing_ID

GROUP BY
Parking_Available,
Power_Backup;
"""

df_parking=pd.read_sql(query,conn)
print(df_parking)
```

```python
#Q9.How does year built influence listing price?
query="""
SELECT
Year_Built,
ROUND(AVG(l.Price),2) AS Avg_Price

FROM listings l
JOIN property_attributes p
ON l.Listing_ID=p.Listing_ID

GROUP BY Year_Built

ORDER BY Year_Built;
"""

df_year=pd.read_sql(query,conn)
print(df_year)
```

```python
#Q10.Which cities have the highest average property prices?
query="""
SELECT

CASE
WHEN Price<100000 THEN 'Below 100K'
WHEN Price BETWEEN 100000 AND 300000 THEN '100K-300K'
WHEN Price BETWEEN 300001 AND 500000 THEN '300K-500K'
ELSE 'Above 500K'
END AS Price_Bucket,

COUNT(*) AS Total_Properties

FROM listings

GROUP BY Price_Bucket;
"""

df_bucket=pd.read_sql(query,conn)
print(df_bucket)
```

```python
#Question 11: What is the average days on market by city?
query = """
SELECT
    l.City,
    ROUND(AVG(s.Days_On_Market),2) AS Avg_Days_On_Market
FROM listings l
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY l.City
ORDER BY Avg_Days_On_Market;
"""

df_avg_days_city = pd.read_sql(query, conn)
print(df_avg_days_city)
```

```python
#Question 12: Which property types sell the fastest?
query = """
SELECT
    l.Property_Type,
    ROUND(AVG(s.Days_On_Market),2) AS Avg_Days_On_Market
FROM listings l
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY l.Property_Type
ORDER BY Avg_Days_On_Market;
"""

df_fast_property = pd.read_sql(query, conn)
print(df_fast_property)
```

```python
#Question 13: What percentage of properties are sold above listing price?
query = """
SELECT
    ROUND(
        100.0 * SUM(
            CASE
                WHEN s.Sale_Price > l.Price THEN 1
                ELSE 0
            END
        ) / COUNT(*), 2
    ) AS Percentage_Sold_Above_Listing
FROM listings l
JOIN sales s
ON l.Listing_ID = s.Listing_ID;
"""

df_above_listing = pd.read_sql(query, conn)
print(df_above_listing)
```

```python
#What is the sale-to-list price ratio by city?
query = """
SELECT
    l.City,
    ROUND(AVG(s.Sale_Price / l.Price),2) AS Sale_To_List_Ratio
FROM listings l
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY l.City
ORDER BY Sale_To_List_Ratio DESC;
"""

df_sale_list_ratio = pd.read_sql(query, conn)
print(df_sale_list_ratio)
```

```python
#Which listings took more than 90 days to sell?
query = """
SELECT
    l.Listing_ID,
    l.City,
    l.Property_Type,
    l.Price,
    s.Sale_Price,
    s.Days_On_Market
FROM listings l
JOIN sales s
ON l.Listing_ID = s.Listing_ID
WHERE s.Days_On_Market > 90
ORDER BY s.Days_On_Market DESC;
"""

df_over_90_days = pd.read_sql(query, conn)
print(df_over_90_days)
```

```python
#Question 16: How does metro distance affect time on market?
query = """
SELECT
    CASE
        WHEN p.Metro_Distance_Km < 1 THEN 'Less than 1 KM'
        WHEN p.Metro_Distance_Km BETWEEN 1 AND 3 THEN '1-3 KM'
        ELSE 'More than 3 KM'
    END AS Metro_Distance,

    ROUND(AVG(s.Days_On_Market),2) AS Avg_Days_On_Market

FROM property_attributes p
JOIN sales s
ON p.Listing_ID = s.Listing_ID

GROUP BY Metro_Distance
ORDER BY Avg_Days_On_Market;
"""

df_metro_days = pd.read_sql(query, conn)
print(df_metro_days)
```

```python
#Question 17: What is the monthly sales trend?
query = """
SELECT
    strftime('%Y-%m', Date_Sold) AS Month,
    COUNT(*) AS Total_Sales
FROM sales
GROUP BY Month
ORDER BY Month;
"""

df_monthly_sales = pd.read_sql(query, conn)
print(df_monthly_sales)
```

```python
#Question 18: Which properties are currently unsold?
query = """
SELECT
    l.*
FROM listings l
LEFT JOIN sales s
ON l.Listing_ID = s.Listing_ID
WHERE s.Listing_ID IS NULL;
"""

df_unsold = pd.read_sql(query, conn)
print(df_unsold)
```

```python
#Question 19: Which agents have closed the most sales?
query = """
SELECT
    a.Agent_ID,
    a.Name,
    COUNT(s.Listing_ID) AS Total_Sales
FROM agents a
JOIN listings l
ON a.Agent_ID = l.Agent_ID
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY a.Agent_ID, a.Name
ORDER BY Total_Sales DESC;
"""

df_agent_sales = pd.read_sql(query, conn)
print(df_agent_sales)
```

```python
#Question 20: Who are the top agents by total sales revenue?
query = """
SELECT
    a.Agent_ID,
    a.Name,
    ROUND(SUM(s.Sale_Price),2) AS Total_Sales_Revenue
FROM agents a
JOIN listings l
ON a.Agent_ID = l.Agent_ID
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY a.Agent_ID, a.Name
ORDER BY Total_Sales_Revenue DESC;
"""

df_agent_revenue = pd.read_sql(query, conn)
print(df_agent_revenue)
```

```python
#Question 21: Which agents close deals the fastest?
query = """
SELECT
    a.Agent_ID,
    a.Name,
    ROUND(AVG(s.Days_on_Market),2) AS Avg_Days_On_Market
FROM agents a
JOIN listings l
ON a.Agent_ID = l.Agent_ID
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY a.Agent_ID, a.Name
ORDER BY Avg_Days_On_Market;
"""

df_fast_agents = pd.read_sql(query, conn)
print(df_fast_agents)
```

```python
#Question 22: Is there a relationship between agent experience and deals closed?
query = """
SELECT
    a.Agent_ID,
    a.Name,
    a.Experience_Years,
    COUNT(s.Listing_ID) AS Deals_Closed
FROM agents a
JOIN listings l
ON a.Agent_ID = l.Agent_ID
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY
    a.Agent_ID,
    a.Name,
    a.Experience_Years
ORDER BY a.Experience_Years DESC;
"""

df_experience_deals = pd.read_sql(query, conn)
print(df_experience_deals)
```

```python
#Question 23: Do higher-rated agents close deals faster?
query = """
SELECT
    a.Rating,
    ROUND(AVG(s.Days_on_Market),2) AS Avg_Days_On_Market
FROM agents a
JOIN listings l
ON a.Agent_ID = l.Agent_ID
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY a.Rating
ORDER BY a.Rating DESC;
"""

df_rating_speed = pd.read_sql(query, conn)
print(df_rating_speed)
```

```python
#Question 24: What is the average commission earned by each agent?
query = """
SELECT
    a.Agent_ID,
    a.Name,
    ROUND(
        AVG(s.Sale_Price * a.Commission_Rate / 100),
        2
    ) AS Avg_Commission
FROM agents a
JOIN listings l
ON a.Agent_ID = l.Agent_ID
JOIN sales s
ON l.Listing_ID = s.Listing_ID
GROUP BY a.Agent_ID, a.Name
ORDER BY Avg_Commission DESC;
"""

df_agent_commission = pd.read_sql(query, conn)
print(df_agent_commission)
```

```python
#Question 25: Which agents have the most active listings?
query = """
SELECT
    a.Agent_ID,
    a.Name,
    COUNT(l.Listing_ID) AS Active_Listings
FROM agents a
JOIN listings l
ON a.Agent_ID = l.Agent_ID
LEFT JOIN sales s
ON l.Listing_ID = s.Listing_ID
WHERE s.Listing_ID IS NULL
GROUP BY a.Agent_ID, a.Name
ORDER BY Active_Listings DESC;
"""

df_active_listings = pd.read_sql(query, conn)
print(df_active_listings)
```

```python
#Question 26: What percentage of buyers are investors versus end users?
query = """
SELECT
    Buyer_Type,
    COUNT(*) AS Total_Buyers,
    ROUND(
        100.0 * COUNT(*) /
        (SELECT COUNT(*) FROM buyers),
        2
    ) AS Percentage
FROM buyers
GROUP BY Buyer_Type;
"""

df_buyer_type = pd.read_sql(query, conn)
print(df_buyer_type)
```

```python
#Question 27: Which cities have the highest loan uptake?
query = """
SELECT
    l.City,
    ROUND(
        100.0 * SUM(
            CASE
                WHEN b.loan_taken = 1 THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS Loan_Uptake_Percentage
FROM buyers b
JOIN sales s
ON b.sale_id = s.Listing_ID
JOIN listings l
ON s.Listing_ID = l.Listing_ID
GROUP BY l.City
ORDER BY Loan_Uptake_Percentage DESC;
"""

df_city_loan = pd.read_sql(query, conn)
print(df_city_loan)
```
